This function creates a secure, random password with a specified length (default is 12). It ensures a mix of character types for better security.

```python
import string
import secrets

def generate_secure_password(length=12):
    if length < 4:
        raise ValueError("Password length should be at least 4 characters.")
    
    # Ensure one character from each category
    categories = [
        secrets.choice(string.ascii_lowercase),
        secrets.choice(string.ascii_uppercase),
        secrets.choice(string.digits),
        secrets.choice(string.punctuation)
    ]
    
    # Fill the rest of the password length
    all_characters = string.ascii_letters + string.digits + string.punctuation
    remaining = [secrets.choice(all_characters) for _ in range(length - 4)]
    
    # Combine and shuffle
    password_list = categories + remaining
    secrets.SystemRandom().shuffle(password_list)
    
    return ''.join(password_list)

# Example usage
print("Generated Secure Password:", generate_secure_password())
```

```python
import string
import secrets

def generate_secure_password(length):
    if length < 4:
        raise ValueError("Password length should be at least 8 characters.")

    # Step 1: Ensure one character from each category
    categories = [
        secrets.choice(string.ascii_lowercase),
        secrets.choice(string.ascii_uppercase),
        secrets.choice(string.digits),
        secrets.choice(string.punctuation)
    ]

    # Step 2: Fill the rest of the password length
    all_characters = string.ascii_letters + string.digits + string.punctuation
    remaining = [secrets.choice(all_characters) for _ in range(length - 4)]

    # Step 3: Combine and shuffle the characters
    password_list = categories + remaining
    secrets.SystemRandom().shuffle(password_list)

    return ''.join(password_list)

# 🔽 Take user input for password length
try:
    user_input = int(input("Enter desired password length (min 4): "))
    password = generate_secure_password(user_input)
    print("Generated Secure Password:", password)
except ValueError as ve:
    print("Error:", ve)
```
